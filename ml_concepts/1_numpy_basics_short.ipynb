```python
import numpy as np
import time
```

## Why NumPy vs Regular Lists?

```python
###########################################################################################
# SPEED: NumPy is 10-100x faster!
python_list = list(range(1000000))
numpy_array = np.arange(1000000)

# Python list - slow
start = time.time()
result = [x * 2 for x in python_list]
print(f"Python list: {time.time() - start:.4f}s")

# NumPy - blazing fast!
start = time.time()
result = numpy_array * 2
print(f"NumPy array: {time.time() - start:.4f}s")


###########################################################################################
# SIMPLICITY: Look at this difference!
list1 = [1, 2, 3]
list2 = [4, 5, 6]

# Python lists - awkward
print("\nPython:", [list1[i] + list2[i] for i in range(len(list1))])

# NumPy - clean & simple
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])
print("NumPy:", arr1 + arr2)
```

## 1. Create Arrays

```python
# From list
arr = np.array([1, 2, 3, 4, 5])
print("Array:", arr)

# Range of numbers
arr_range = np.arange(0, 10, 2)
print("Range:", arr_range)

# 2D array
matrix = np.array([[1, 2, 3], [4, 5, 6]])
print("Matrix:\n", matrix)
```

## 2. Array Operations

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# Math operations
print("Add:", a + b)
print("Multiply:", a * b)
print("Square:", a ** 2)
```

## 3. Useful Statistical Functions

```python
data = np.array([10, 20, 30, 40, 50])

print("Mean:", data.mean())
print("Max:", data.max())
print("Sum:", data.sum())
print("Shape:", data.shape)
```

## 4. Reshape Arrays

```python
# 1D to 2D
arr = np.arange(1, 13)
print("before reshape:\n", arr)
reshaped = arr.reshape(3, 4)
print("Reshaped:\n", reshaped)
```

## 5. Indexing & Slicing

```python
arr = np.array([10, 20, 30, 40, 50])

print("First element:", arr[0])
print("Last element:", arr[-1])
print("Slice [1:4]:", arr[1:4])
print("Elements > 25:", arr[arr > 25])
```

# Advanced NumPy (Bonus!)

## 6. Broadcasting - Auto Size Matching!

```python
# Add a single number to entire array!
arr = np.array([[1, 2, 3], [4, 5, 6]])
print("Original:\n", arr)
print("\nAdd 10 to all:\n", arr + 10)

# Multiply row-wise
row_multiplier = np.array([10, 100, 1000])
print("\nRow multiply:\n", arr * row_multiplier)
```

## 7. Vectorized Operations - No Loops!

```python
# Apply complex math to millions of numbers - no loops!
angles = np.linspace(0, 2*np.pi, 5)
print("Angles:", angles)
print("Sine:", np.sin(angles))
print("Cosine:", np.cos(angles))

# Conditional operations
scores = np.array([45, 78, 92, 65, 88])
grades = np.where(scores >= 75, 'Pass', 'Fail')
print("\nScores:", scores)
print("Grades:", grades)
```

## 8. Matrix Operations - Linear Algebra!

```python
# Matrix multiplication (used in AI/ML!)
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("Matrix A:\n", A)
print("\nMatrix B:\n", B)
print("\nDot Product:\n", np.dot(A, B))
print("\nTranspose:\n", A.T)
```

## 9. Random Numbers - Simulations!

```python
# Generate random data instantly
np.random.seed(42)  # For reproducibility

# Random integers
print("Random integers:", np.random.randint(1, 100, size=5))

# Random floats (0 to 1)
print("Random floats:", np.random.random(5))

# Normal distribution (mean=0, std=1)
print("Normal dist:", np.random.randn(5))

# Random choice
fruits = np.array(['Apple', 'Banana', 'Orange', 'Mango'])
print("Random fruit:", np.random.choice(fruits))
```

## 10. Stacking & Splitting Arrays

```python
# Combine arrays
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("Vertical stack:\n", np.vstack([a, b]))
print("\nHorizontal stack:", np.hstack([a, b]))

# Split arrays
arr = np.arange(12)
print("\nOriginal:", arr)
split = np.split(arr, 3)  # Split into 3 parts
print("Split into 3:", split)
```

## 11. Aggregations Along Axis

```python
# Work with specific dimensions
sales = np.array([[100, 150, 200],  # Store 1
                  [120, 180, 210],  # Store 2
                  [90, 140, 190]])  # Store 3

print("Sales data:\n", sales)
print("\nTotal per store (row sum):", sales.sum(axis=1))
print("Total per month (column sum):", sales.sum(axis=0))
print("Average per month:", sales.mean(axis=0))
```

## 12. Unique & Sorting

```python
# Find unique values
arr = np.array([1, 2, 2, 3, 3, 3, 4, 4, 4, 4])
print("Original:", arr)
print("Unique values:", np.unique(arr))
unique, counts = np.unique(arr, return_counts=True)
print("Value counts:", dict(zip(unique, counts)))

# Sort array
unsorted = np.array([5, 2, 8, 1, 9])
print("\nUnsorted:", unsorted)
print("Sorted:", np.sort(unsorted))
print("Sort indices:", np.argsort(unsorted))
```

## 13. Real-World Example: Image Processing!

```python
# Images are just arrays! (Height x Width x RGB)
# Simulate a small 5x5 RGB image
image = np.random.randint(0, 256, size=(5, 5, 3))
print("Image shape (H x W x Channels):", image.shape)

# Grayscale conversion (average RGB channels)
grayscale = image.mean(axis=2)
print("\nGrayscale shape:", grayscale.shape)

# Brightness adjustment (add 50 to all pixels)
brighter = np.clip(image + 50, 0, 255)  # Keep values in 0-255
print("Brightened shape:", brighter.shape)

print("\nThis is how PIL, OpenCV work under the hood!")
```
