```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime, timedelta

plt.style.use('dark_background')
np.random.seed(42)

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
pd.set_option('display.precision', 2)
```

## 1. Quick Data Exploration (5-Second Rule)

```python
# Create sample sales dataset
np.random.seed(42)
n = 1000

dates = pd.date_range('2023-01-01', periods=n, freq='D')
df = pd.DataFrame({
    'date': np.random.choice(dates, n),
    'product': np.random.choice(['Laptop', 'Phone', 'Tablet', 'Monitor', 'Keyboard'], n),
    'region': np.random.choice(['North', 'South', 'East', 'West'], n),
    'sales': np.random.randint(100, 5000, n),
    'quantity': np.random.randint(1, 20, n),
    'customer_age': np.random.randint(18, 70, n),
    'discount': np.random.choice([0, 5, 10, 15, 20], n)
})

# Add some missing values
df.loc[np.random.choice(df.index, 50, replace=False), 'discount'] = np.nan
df.loc[np.random.choice(df.index, 30, replace=False), 'customer_age'] = np.nan

print("TRICK 1: ONE-LINE DATA EXPLORATION")
print("="*70)
print("\n📊 QUICK SNAPSHOT (The 5-Second Rule):")
print("\n1️⃣ Shape: How much data?")
print(f"   df.shape → {df.shape}  (1000 rows × 7 columns)")

print("\n2️⃣ Info: Data types + missing values")
print("   df.info() →")
df.info()

print("\n3️⃣ Describe: Statistics in one shot")
print("   df.describe() →")
print(df.describe())

print("\n4️⃣ Value Counts: Distribution of categories")
print("   df['product'].value_counts() →")
print(df['product'].value_counts())

print("\n5️⃣ Missing Data: Quick check")
print("   df.isnull().sum() →")
print(df.isnull().sum())

print("\n💡 PRO TIP: Run these 5 commands FIRST!")
print("   They give you 80% understanding in 5 seconds.")
print("="*70)
```

## 2. Smart Filtering (3 Ways)

```python
print("\nTRICK 2: FILTERING DATA LIKE A PRO")
print("="*70)

# Method 1: Boolean Indexing (Classic)
print("\n1️⃣ BOOLEAN INDEXING (Most Common):")
high_sales = df[df['sales'] > 3000]
print(f"   df[df['sales'] > 3000] → {len(high_sales)} rows")

# Multiple conditions
print("\n   Multiple conditions with & (and) | (or):")
filtered = df[(df['sales'] > 3000) & (df['region'] == 'North')]
print(f"   High sales in North → {len(filtered)} rows")

# Method 2: Query (Cleaner for complex filters)
print("\n2️⃣ QUERY METHOD (Cleaner Syntax):")
query_result = df.query('sales > 3000 and region == "North"')
print(f"   df.query('sales > 3000 and region == \"North\"') → {len(query_result)} rows")
print("   ✓ More readable!")
print("   ✓ No brackets needed")

# Method 3: isin (Check membership)
print("\n3️⃣ ISIN METHOD (Check if value in list):")
selected_products = df[df['product'].isin(['Laptop', 'Phone'])]
print(f"   df[df['product'].isin(['Laptop', 'Phone'])] → {len(selected_products)} rows")
print("   ✓ Better than (df['product'] == 'Laptop') | (df['product'] == 'Phone')")

# NOT in list
print("\n   NOT in list:")
not_in = df[~df['product'].isin(['Keyboard'])]
print(f"   df[~df['product'].isin(['Keyboard'])] → {len(not_in)} rows")

# Practical example: Complex filter
print("\n🎯 PRACTICAL EXAMPLE (Complex Filter):")
result = df.query('sales > 2000 and region in ["North", "South"] and quantity >= 5')
print(f"   High-value sales in North/South with quantity ≥ 5 → {len(result)} rows")
print("="*70)

# Show sample
print("\nSample of filtered data:")
print(result.head())
```

## 3. Missing Data Mastery (3 Essential Methods)

```python
print("\nTRICK 3: HANDLING MISSING DATA")
print("="*70)

# Check missing data
print("\n📊 MISSING DATA SUMMARY:")
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)
missing_df = pd.DataFrame({
    'Missing Count': missing,
    'Percentage': missing_pct
})
print(missing_df[missing_df['Missing Count'] > 0])

# Method 1: Fill with value
print("\n1️⃣ FILLNA (Replace with value):")
df_filled = df.copy()
df_filled['discount'] = df_filled['discount'].fillna(0)
print("   df['discount'].fillna(0) → Replace NaN with 0")
print(f"   Before: {df['discount'].isnull().sum()} missing")
print(f"   After: {df_filled['discount'].isnull().sum()} missing ✓")

# Fill with mean/median
print("\n   Fill with MEAN/MEDIAN:")
df_filled['customer_age'] = df_filled['customer_age'].fillna(df_filled['customer_age'].median())
print(f"   df['customer_age'].fillna(df['customer_age'].median()) → {df_filled['customer_age'].isnull().sum()} missing ✓")

# Method 2: Forward/Backward fill
print("\n2️⃣ FFILL / BFILL (Forward/Backward Fill):")
df_ffill = df.copy()
df_ffill['discount'] = df_ffill['discount'].fillna(method='ffill')
print("   df['discount'].fillna(method='ffill') → Use previous value")
print("   ✓ Good for time-series data!")

# Method 3: Drop missing
print("\n3️⃣ DROPNA (Remove rows with missing):")
df_dropped = df.dropna()
print(f"   df.dropna() → {len(df)} rows → {len(df_dropped)} rows")
print(f"   Lost {len(df) - len(df_dropped)} rows ({(len(df) - len(df_dropped))/len(df)*100:.1f}%)")

# Drop if specific column is null
print("\n   Drop if specific column is null:")
df_dropped_subset = df.dropna(subset=['discount'])
print(f"   df.dropna(subset=['discount']) → {len(df_dropped_subset)} rows")

print("\n💡 PRO TIP: Choose method based on data:")
print("   - Categorical: fillna(mode) or fillna('Unknown')")
print("   - Numerical: fillna(mean) or fillna(median)")
print("   - Time-series: ffill() or bfill()")
print("   - Too many missing: dropna()")
print("="*70)
```

## 4. Column Magic (Apply, Map, Lambda)

```python
print("\nTRICK 4: CREATING & TRANSFORMING COLUMNS")
print("="*70)

# Simple column creation
print("\n1️⃣ SIMPLE ARITHMETIC:")
df['revenue'] = df['sales'] * df['quantity']
print("   df['revenue'] = df['sales'] * df['quantity']")
print(f"   New column created! Sample: {df['revenue'].head(3).tolist()}")

# Apply with lambda
print("\n2️⃣ APPLY + LAMBDA (Row-wise operations):")
df['sales_category'] = df['sales'].apply(lambda x: 'High' if x > 3000 else 'Medium' if x > 1500 else 'Low')
print("   df['sales'].apply(lambda x: 'High' if x > 3000 else ...)")
print(f"   Distribution: {df['sales_category'].value_counts().to_dict()}")

# Apply with function
print("\n3️⃣ APPLY WITH FUNCTION (Complex logic):")
def calculate_commission(row):
    if row['sales'] > 3000:
        return row['sales'] * 0.10  # 10% commission
    elif row['sales'] > 1500:
        return row['sales'] * 0.05  # 5% commission
    else:
        return row['sales'] * 0.02  # 2% commission

df['commission'] = df.apply(calculate_commission, axis=1)
print("   df.apply(calculate_commission, axis=1)")
print(f"   Commission range: ${df['commission'].min():.0f} - ${df['commission'].max():.0f}")

# Map (For replacements)
print("\n4️⃣ MAP (Dictionary-based replacement):")
region_map = {'North': 'N', 'South': 'S', 'East': 'E', 'West': 'W'}
df['region_code'] = df['region'].map(region_map)
print("   df['region'].map({'North': 'N', 'South': 'S', ...})")
print(f"   Sample mapping: {df[['region', 'region_code']].head(3).to_dict('records')}")

# Vectorized operations (FASTEST!)
print("\n5️⃣ VECTORIZED (NumPy - FASTEST!):")
df['discount_amount'] = df['sales'] * (df['discount'].fillna(0) / 100)
print("   df['sales'] * (df['discount'] / 100) → Vectorized!")
print("   ✓ 100x faster than apply for simple math!")

print("\n💡 PERFORMANCE TIP:")
print("   Vectorized (fastest) > apply(function) > apply(lambda)")
print("   Use vectorized whenever possible!")
print("="*70)

# Show sample
print("\nSample with new columns:")
print(df[['sales', 'quantity', 'revenue', 'sales_category', 'commission']].head())
```

## 5. GroupBy Secrets (Split-Apply-Combine)

```python
print("\nTRICK 5: GROUPBY MASTERY")
print("="*70)

# Basic groupby
print("\n1️⃣ BASIC GROUPBY (Single aggregation):")
region_sales = df.groupby('region')['sales'].sum()
print("   df.groupby('region')['sales'].sum()")
print(region_sales)

# Multiple aggregations
print("\n2️⃣ MULTIPLE AGGREGATIONS (agg):")
stats = df.groupby('product')['sales'].agg(['count', 'sum', 'mean', 'min', 'max'])
print("   df.groupby('product')['sales'].agg(['count', 'sum', 'mean', 'min', 'max'])")
print(stats)

# Different functions for different columns
print("\n3️⃣ DIFFERENT FUNCTIONS PER COLUMN:")
multi_agg = df.groupby('region').agg({
    'sales': ['sum', 'mean'],
    'quantity': ['sum', 'mean'],
    'customer_age': 'mean'
})
print("   df.groupby('region').agg({'sales': ['sum', 'mean'], ...})")
print(multi_agg)

# Named aggregations (Clean column names!)
print("\n4️⃣ NAMED AGGREGATIONS (Clean output):")
summary = df.groupby('product').agg(
    total_sales=('sales', 'sum'),
    avg_sales=('sales', 'mean'),
    total_quantity=('quantity', 'sum'),
    num_transactions=('sales', 'count')
).round(2)
print("   df.groupby('product').agg(total_sales=('sales', 'sum'), ...)")
print(summary)

# Group by multiple columns
print("\n5️⃣ MULTIPLE GROUP COLUMNS:")
product_region = df.groupby(['product', 'region'])['sales'].sum().unstack(fill_value=0)
print("   df.groupby(['product', 'region'])['sales'].sum().unstack()")
print(product_region)

# Transform (keep original size)
print("\n6️⃣ TRANSFORM (Add group stats to original df):")
df['region_avg_sales'] = df.groupby('region')['sales'].transform('mean')
print("   df.groupby('region')['sales'].transform('mean')")
print("   ✓ Adds group average back to each row!")
print(f"   Sample: {df[['region', 'sales', 'region_avg_sales']].head(3).to_dict('records')}")

print("\n💡 PRO TIP: Named aggregations keep code readable!")
print("   Avoid multi-level column names - use named agg instead.")
print("="*70)
```

## 6. Date Magic (Parse, Extract, Resample)

```python
print("\nTRICK 6: DATE/TIME OPERATIONS")
print("="*70)

# Ensure date is datetime
df['date'] = pd.to_datetime(df['date'])

# Extract date parts
print("\n1️⃣ EXTRACT DATE PARTS:")
df['year'] = df['date'].dt.year
df['month'] = df['date'].dt.month
df['day'] = df['date'].dt.day
df['day_name'] = df['date'].dt.day_name()
df['week'] = df['date'].dt.isocalendar().week
print("   df['date'].dt.year, .dt.month, .dt.day_name(), .dt.week")
print(f"   Sample: {df[['date', 'year', 'month', 'day_name', 'week']].head(3).to_dict('records')}")

# Date filtering
print("\n2️⃣ DATE FILTERING:")
march_data = df[df['date'].dt.month == 3]
print(f"   df[df['date'].dt.month == 3] → {len(march_data)} rows in March")

# Date range
date_range = df[(df['date'] >= '2023-06-01') & (df['date'] <= '2023-06-30')]
print(f"   June 2023 data → {len(date_range)} rows")

# Resample (Time-series aggregation)
print("\n3️⃣ RESAMPLE (Time-series aggregation):")
df_sorted = df.sort_values('date')
df_sorted = df_sorted.set_index('date')

# Monthly aggregation
monthly = df_sorted['sales'].resample('M').sum()
print("   df.resample('M').sum() → Monthly totals")
print(monthly.head())

# Weekly aggregation
print("\n   Weekly aggregation:")
weekly = df_sorted['sales'].resample('W').agg(['sum', 'mean', 'count'])
print("   df.resample('W').agg(['sum', 'mean', 'count'])")
print(weekly.head())

# Reset index
df = df_sorted.reset_index()

# Date difference
print("\n4️⃣ DATE CALCULATIONS:")
min_date = df['date'].min()
df['days_since_start'] = (df['date'] - min_date).dt.days
print(f"   Days since start: 0 to {df['days_since_start'].max()} days")

print("\n💡 PRO TIP: Use dt accessor for date operations!")
print("   dt.year, dt.month, dt.day_name(), dt.week, etc.")
print("="*70)
```

## 7. String Operations (Split, Extract, Clean)

```python
print("\nTRICK 7: STRING OPERATIONS")
print("="*70)

# Create sample data with strings
df['product_full'] = df['product'] + ' - ' + df['region'] + ' - ' + df['sales_category']

print("\n1️⃣ STRING METHODS (str accessor):")

# Lowercase/Uppercase
df['product_lower'] = df['product'].str.lower()
df['product_upper'] = df['product'].str.upper()
print("   df['product'].str.lower() / .str.upper()")
print(f"   Sample: {df[['product', 'product_lower', 'product_upper']].head(2).to_dict('records')}")

# Contains
print("\n2️⃣ STRING FILTERING (contains):")
laptops = df[df['product'].str.contains('Laptop', case=False)]
print(f"   df[df['product'].str.contains('Laptop')] → {len(laptops)} rows")

# Starts with / Ends with
print("\n3️⃣ STARTS/ENDS WITH:")
starts_p = df[df['product'].str.startswith('P')]
print(f"   df['product'].str.startswith('P') → {len(starts_p)} rows (Phone)")

# Split string
print("\n4️⃣ SPLIT STRING:")
split_result = df['product_full'].str.split(' - ', expand=True)
split_result.columns = ['product_split', 'region_split', 'category_split']
print("   df['product_full'].str.split(' - ', expand=True)")
print(split_result.head(3))

# Extract with regex
print("\n5️⃣ EXTRACT (Regex):")
# Example: Extract numbers if they existed in string
df['product_length'] = df['product'].str.len()
print("   df['product'].str.len() → String length")
print(f"   Sample: {df[['product', 'product_length']].head(3).to_dict('records')}")

# Replace
print("\n6️⃣ REPLACE:")
df['product_clean'] = df['product'].str.replace('Laptop', 'Notebook')
print("   df['product'].str.replace('Laptop', 'Notebook')")
print(f"   Sample: {df[df['product'] == 'Laptop'][['product', 'product_clean']].head(2).to_dict('records')}")

# Strip whitespace
print("\n7️⃣ STRIP (Remove whitespace):")
print("   df['column'].str.strip() / .str.lstrip() / .str.rstrip()")
print("   ✓ Essential for cleaning data!")

print("\n💡 PRO TIP: Always use str accessor for string operations!")
print("   Never: df['col'].apply(lambda x: x.lower())")
print("   Always: df['col'].str.lower() → 100x faster!")
print("="*70)
```

## 8. Method Chaining (Write Clean Code)

```python
print("\nTRICK 8: METHOD CHAINING (Clean, Readable Code)")
print("="*70)

# Bad: Multiple steps
print("\n❌ BAD (Multiple intermediate variables):")
print("""
df_temp1 = df[df['sales'] > 2000]
df_temp2 = df_temp1.groupby('region')['sales'].sum()
df_temp3 = df_temp2.sort_values(ascending=False)
result = df_temp3.head(3)
""")

# Good: Method chaining
print("\n✅ GOOD (Method chaining):")
result = (
    df[df['sales'] > 2000]
    .groupby('region')['sales']
    .sum()
    .sort_values(ascending=False)
    .head(3)
)
print("""
result = (
    df[df['sales'] > 2000]
    .groupby('region')['sales']
    .sum()
    .sort_values(ascending=False)
    .head(3)
)
""")
print("Result:")
print(result)

# Complex chain example
print("\n🎯 COMPLEX EXAMPLE:")
summary = (
    df
    .query('sales > 1000')
    .assign(
        revenue=lambda x: x['sales'] * x['quantity'],
        profit_margin=lambda x: (x['sales'] - x['discount_amount']) / x['sales']
    )
    .groupby('product')
    .agg(
        total_revenue=('revenue', 'sum'),
        avg_margin=('profit_margin', 'mean'),
        count=('sales', 'count')
    )
    .sort_values('total_revenue', ascending=False)
    .reset_index()
)

print("Summary:")
print(summary)

print("\n💡 PRO TIPS:")
print("   1. Use parentheses for multi-line chains")
print("   2. One method per line → Readable!")
print("   3. Use .assign() to create columns in chains")
print("   4. .pipe() for custom functions in chains")
print("="*70)
```

## 9. Performance Tips (Speed Hacks)

```python
print("\nTRICK 9: PERFORMANCE OPTIMIZATION")
print("="*70)

print("\n1️⃣ AVOID LOOPS (Use vectorization):")
print("\n❌ SLOW:")
print("""
for i in range(len(df)):
    df.loc[i, 'new_col'] = df.loc[i, 'sales'] * 2
""")

print("\n✅ FAST (100x faster!):")
print("df['new_col'] = df['sales'] * 2  # Vectorized!")

print("\n2️⃣ USE CATEGORY DTYPE (Save memory):")
print(f"\nBefore: {df['product'].memory_usage(deep=True) / 1024:.2f} KB")
df['product'] = df['product'].astype('category')
print(f"After (category): {df['product'].memory_usage(deep=True) / 1024:.2f} KB")
print("✓ Use for columns with few unique values!")

print("\n3️⃣ READ CSV EFFICIENTLY:")
print("""
# Specify dtypes
df = pd.read_csv('data.csv', dtype={'column': 'category'})

# Read only needed columns
df = pd.read_csv('data.csv', usecols=['col1', 'col2'])

# Parse dates while reading
df = pd.read_csv('data.csv', parse_dates=['date_column'])

# Read in chunks (for huge files)
for chunk in pd.read_csv('data.csv', chunksize=10000):
    process(chunk)
""")

print("\n4️⃣ USE .loc and .iloc (Avoid chained indexing):")
print("\n❌ BAD (Chained indexing):")
print("df[df['sales'] > 2000]['region'] = 'High'  # Warning!")

print("\n✅ GOOD:")
print("df.loc[df['sales'] > 2000, 'region'] = 'High'  # Correct!")

print("\n5️⃣ QUERY vs BOOLEAN (Query is faster!):")
print("\nFor complex filters, query() is 2-3x faster:")
print("df.query('sales > 2000 and region == \"North\"')  # Faster!")

print("\n💡 GOLDEN RULES:")
print("   1. NEVER use loops → Use vectorization")
print("   2. Use 'category' dtype for low-cardinality columns")
print("   3. Specify dtypes when reading CSV")
print("   4. Use .loc / .iloc for assignments")
print("   5. query() for complex filters")
print("="*70)
```

## 10. Export Shortcuts (Save Your Work)

```python
print("\nTRICK 10: EXPORT DATA EFFICIENTLY")
print("="*70)

print("\n1️⃣ TO_CSV (With options):")
print("""
# Basic
df.to_csv('output.csv', index=False)  # Don't save index

# Custom separator
df.to_csv('output.tsv', sep='\t', index=False)  # Tab-separated

# Specific columns
df[['product', 'sales']].to_csv('summary.csv', index=False)

# With encoding
df.to_csv('output.csv', encoding='utf-8', index=False)
""")

print("\n2️⃣ TO_EXCEL (Multiple sheets):")
print("""
# Single sheet
df.to_excel('output.xlsx', sheet_name='Sales', index=False)

# Multiple sheets
with pd.ExcelWriter('output.xlsx') as writer:
    df_north.to_excel(writer, sheet_name='North', index=False)
    df_south.to_excel(writer, sheet_name='South', index=False)
    summary.to_excel(writer, sheet_name='Summary', index=False)
""")

print("\n3️⃣ TO_PARQUET (For large data):")
print("""
# Parquet is 10x smaller and faster than CSV!
df.to_parquet('output.parquet')

# Read back
df = pd.read_parquet('output.parquet')

✓ Preserves data types!
✓ Compressed automatically
✓ 10-100x faster than CSV
""")

print("\n4️⃣ TO_JSON (For APIs):")
print("""
# Default (records)
df.to_json('output.json', orient='records')

# Pretty print
df.to_json('output.json', orient='records', indent=2)

# Split format
df.to_json('output.json', orient='split')
""")

print("\n5️⃣ TO_CLIPBOARD (Quick copy):")
print("""
df.head(10).to_clipboard(index=False)
# Now paste into Excel or Google Sheets!
""")

print("\n💡 PRO TIPS:")
print("   - CSV: Universal, but slow for large data")
print("   - Excel: Business users, multiple sheets")
print("   - Parquet: Large data, preserves types, FAST!")
print("   - JSON: APIs, web apps")
print("   - Clipboard: Quick sharing (small data)")
print("="*70)

# Practical example
print("\n🎯 PRACTICAL: Save summary reports")
print("\nCreating region summary...")
region_summary = (
    df.groupby('region')
    .agg(
        total_sales=('sales', 'sum'),
        avg_sales=('sales', 'mean'),
        transactions=('sales', 'count')
    )
    .round(2)
)
print(region_summary)

print("\n✓ In real scenario, save with: region_summary.to_excel('region_report.xlsx')")
```
